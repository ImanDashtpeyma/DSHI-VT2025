# Student Alcohol Consumption Dataset Analysis

## 1. Dataset Source
Source: https://www.kaggle.com/datasets/uciml/student-alcohol-consumption/data

## 2. Domain Area
The dataset was collected from two secondary schools in Portugal through school reports and questionnaires. It aims to understand the relationship between students' social, demographic, and school-related features with their academic performance and alcohol consumption patterns. The data was gathered to analyze factors affecting student achievement in secondary education.

## 3. Data File Structure
The dataset consists of two separate .csv files (student-mat.csv and student-por.csv) for Mathematics and Portuguese language students respectively. The files are semicolon-separated (;) and include headers. We will use both files for comprehensive analysis, merging them appropriately to avoid duplicate student records.

## 4. Features and Observations
- Mathematics dataset: 395 observations
- Portuguese dataset: 649 observations
- Total features: 33 variables per dataset

## 5. Numerical Features (8 features)
- age: 15-22
- Medu, Fedu: Mother's and Father's education (0-4)
- failures: Past class failures (0-4)
- absences: Number of school absences (0-93)
- G1, G2, G3: Grades (0-20)
- Dalc, Walc: Workday/Weekend alcohol consumption (1-5)

## 6. Categorical Features (15 features)
Nominal:
- school: 'GP' or 'MS'
- Mjob, Fjob: Mother's/Father's job
- reason: Reason to choose school
- guardian: Student's guardian

Ordinal:
- traveltime: Home to school travel time (1-4)
- studytime: Weekly study time (1-4)
- famrel: Quality of family relationships (1-5)
- freetime: Free time after school (1-5)
- goout: Going out with friends (1-5)

## 7. Binary Features (8 features)
- sex: 'F' or 'M'
- address: 'U' (urban) or 'R' (rural)
- famsize: 'LE3' or 'GT3'
- Pstatus: 'T' or 'A'
- schoolsup: 'yes' or 'no'
- famsup: 'yes' or 'no'
- paid: 'yes' or 'no'
- internet: 'yes' or 'no'

About Dataset
Context:
The data were obtained in a survey of students math and portuguese language courses in secondary school. It contains a lot of interesting social, gender and study information about students. You can use it for some EDA or try to predict students final grade.
Content:
Attributes for both student-mat.csv (Math course) and student-por.csv (Portuguese language course) datasets:

11.	reason - reason to choose this school (nominal: close to 'home', school 'reputation', 'course' preference or 'other')
12.	guardian - student's guardian (nominal: 'mother', 'father' or 'other')
13.	traveltime - home to school travel time (numeric: 1 - <15 min., 2 - 15 to 30 min., 3 - 30 min. to 1 hour, or 4 - >1 hour)
14.	studytime - weekly study time (numeric: 1 - <2 hours, 2 - 2 to 5 hours, 3 - 5 to 10 hours, or 4 - >10 hours)
15.	failures - number of past class failures (numeric: n if 1<=n<3, else 4)
16.	schoolsup - extra educational support (binary: yes or no)
17.	famsup - family educational support (binary: yes or no)
18.	paid - extra paid classes within the course subject (Math or Portuguese) (binary: yes or no)
19.	activities - extra-curricular activities (binary: yes or no)
20.	nursery - attended nursery school (binary: yes or no)
21.	higher - wants to take higher education (binary: yes or no)
22.	internet - Internet access at home (binary: yes or no)
23.	romantic - with a romantic relationship (binary: yes or no)
24.	famrel - quality of family relationships (numeric: from 1 - very bad to 5 - excellent)
25.	freetime - free time after school (numeric: from 1 - very low to 5 - very high)
26.	goout - going out with friends (numeric: from 1 - very low to 5 - very high)
27.	Dalc - workday alcohol consumption (numeric: from 1 - very low to 5 - very high)
28.	Walc - weekend alcohol consumption (numeric: from 1 - very low to 5 - very high)
29.	health - current health status (numeric: from 1 - very bad to 5 - very good)
30.	absences - number of school absences (numeric: from 0 to 93)
These grades are related with the course subject, Math or Portuguese:
1.	G1 - first period grade (numeric: from 0 to 20)

```python
#import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
import pandas as pd

# Load the datasets
math_df = pd.read_csv("./Data/student-mat.csv")
portuguese_df = pd.read_csv("./Data/student-por.csv")

# Add a new column 'course' to each dataset
math_df['course'] = 'mat'
portuguese_df['course'] = 'por'

# Combine the two datasets
combined_df = pd.concat([math_df, portuguese_df])

# Save the combined dataset to a new CSV file
combined_df.to_csv("./Data/combined_student_data.csv", index=False)

print("Combined dataset saved as 'combined_student_data.csv'")
```

### Data Preprocessing Pipeline

- Special focus on alcohol consumption (Dalc, Walc) and academic performance (G1, G2, G3)

```python
sns.set_theme(style='darkgrid')
# Set style for better visualization
```

```python
# Read the data
data = pd.read_csv('./Data/combined_student_data.csv')

# Display the first few rows
print("First few rows of the dataset:")
data.head()
```

```python
# Basic information about alcohol consumption variables
# Dalc: workday alcohol consumption
# Walc: weekend alcohol consumption
print("\nAlcohol Consumption Statistics:")
alcohol_stats = data[['Dalc', 'Walc']].describe()
alcohol_stats
```

```python
# Visualize alcohol consumption patterns
plt.figure(figsize=(12, 5))

plt.subplot(121)
sns.countplot(data=data, x='Dalc')
plt.title('Workday Alcohol Consumption')
plt.xlabel('Consumption Level (1-5)')

plt.subplot(122)
sns.countplot(data=data, x='Walc')
plt.title('Weekend Alcohol Consumption')
plt.xlabel('Consumption Level (1-5)')

plt.tight_layout()
plt.show()
```

```python
# Relationship between workday and weekend alcohol consumption
plt.figure(figsize=(10, 6))
sns.heatmap(pd.crosstab(data['Dalc'], data['Walc']), 
            annot=True, 
            fmt='d', 
            cmap='YlOrRd')
plt.title('Workday vs Weekend Alcohol Consumption')
plt.xlabel('Weekend Alcohol Consumption')
plt.ylabel('Workday Alcohol Consumption')
plt.show()
```

```python
# Impact of alcohol consumption on final grades
plt.figure(figsize=(15, 5))

plt.subplot(121)
sns.boxplot(x='Dalc', y='G3', data=data)
plt.title('Workday Alcohol Consumption vs Final Grade')
plt.xlabel('Workday Alcohol Consumption Level')
plt.ylabel('Final Grade')

plt.subplot(122)
sns.boxplot(x='Walc', y='G3', data=data)
plt.title('Weekend Alcohol Consumption vs Final Grade')
plt.xlabel('Weekend Alcohol Consumption Level')
plt.ylabel('Final Grade')

plt.tight_layout()
plt.show()
```

```python
# Average grades by alcohol consumption levels
print("Average Final Grades by Alcohol Consumption Levels:")
print("\nWorkday Alcohol Consumption:")
print(data.groupby('Dalc')['G3'].mean())
print("\nWeekend Alcohol Consumption:")
print(data.groupby('Walc')['G3'].mean())
```

```python
# Gender differences in alcohol consumption
plt.figure(figsize=(12, 5))

plt.subplot(121)
sns.boxplot(x='sex', y='Dalc', data=data)
plt.title('Workday Alcohol Consumption by Gender')

plt.subplot(122)
sns.boxplot(x='sex', y='Walc', data=data)
plt.title('Weekend Alcohol Consumption by Gender')

plt.tight_layout()
plt.show()
```

```python
# Correlation analysis with alcohol consumption
variables = ['age', 'Medu', 'Fedu', 'studytime', 'failures', 
            'goout', 'Dalc', 'Walc', 'absences', 'G1', 'G2', 'G3']

plt.figure(figsize=(12, 8))
correlation = data[variables].corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Going out and alcohol consumption relationship
plt.figure(figsize=(10, 6))
sns.boxplot(x='goout', y='Walc', data=data)
plt.title('Going Out Frequency vs Weekend Alcohol Consumption')
plt.xlabel('Going Out Frequency (1-5)')
plt.ylabel('Weekend Alcohol Consumption Level')
plt.show()
```

```python
# Study time vs alcohol consumption
plt.figure(figsize=(12, 5))

plt.subplot(121)
sns.boxplot(x='studytime', y='Dalc', data=data)
plt.title('Study Time vs Workday Alcohol Consumption')
plt.xlabel('Study Time (1-4)')

plt.subplot(122)
sns.boxplot(x='studytime', y='Walc', data=data)
plt.title('Study Time vs Weekend Alcohol Consumption')
plt.xlabel('Study Time (1-4)')

plt.tight_layout()
plt.show()
```

```python
# Create a combined alcohol consumption score
data['total_alc'] = data['Dalc'] + data['Walc']

# Academic performance by total alcohol consumption
plt.figure(figsize=(10, 6))
sns.boxplot(x='total_alc', y='G3', data=data)
plt.title('Total Alcohol Consumption vs Final Grade')
plt.xlabel('Total Alcohol Consumption Score')
plt.ylabel('Final Grade')
plt.show()
```

```python
# Statistical summary of grades by alcohol consumption levels
print("Statistical Summary of Grades by Alcohol Consumption:")
print("\nHigh Alcohol Consumption (Total Score > 5):")
high_alc = data[data['total_alc'] > 5]['G3'].describe()
print("\nLow Alcohol Consumption (Total Score <= 5):")
low_alc = data[data['total_alc'] <= 5]['G3'].describe()

pd.DataFrame({
    'High_Alcohol': high_alc,
    'Low_Alcohol': low_alc
})
```

```python
# Additional factors analysis
factors = ['health', 'absences', 'failures']
plt.figure(figsize=(15, 5))

for i, factor in enumerate(factors, 1):
    plt.subplot(1, 3, i)
    sns.scatterplot(data=data, x='total_alc', y=factor)
    plt.title(f'Alcohol Consumption vs {factor.capitalize()}')

plt.tight_layout()
plt.show()
```

```python
# Final summary statistics
summary = pd.DataFrame({
    'Avg_Daily_Alcohol': data['Dalc'].mean(),
    'Avg_Weekend_Alcohol': data['Walc'].mean(),
    'High_Alcohol_Students_%': (data['total_alc'] > 5).mean() * 100,
    'Correlation_Alcohol_Grades': data[['total_alc', 'G3']].corr().iloc[0,1]
}, index=['Statistics'])

summary.round(2)
```
